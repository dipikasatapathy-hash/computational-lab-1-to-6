```python
# Crossword Puzzle Generator using Backtracking

SIZE = 10


# Create an empty crossword grid
def create_grid():
    return [['.' for _ in range(SIZE)] for _ in range(SIZE)]


# Check whether a word can be placed horizontally
def can_place_horizontal(grid, word, row, col):
    if col + len(word) > SIZE:
        return False

    for i in range(len(word)):
        if grid[row][col + i] not in ('.', word[i]):
            return False

    return True


# Check whether a word can be placed vertically
def can_place_vertical(grid, word, row, col):
    if row + len(word) > SIZE:
        return False

    for i in range(len(word)):
        if grid[row + i][col] not in ('.', word[i]):
            return False

    return True


# Place a word horizontally
def place_horizontal(grid, word, row, col):
    for i in range(len(word)):
        grid[row][col + i] = word[i]


# Place a word vertically
def place_vertical(grid, word, row, col):
    for i in range(len(word)):
        grid[row + i][col] = word[i]


# Remove a word horizontally
def remove_horizontal(grid, word, row, col):
    for i in range(len(word)):
        grid[row][col + i] = '.'


# Remove a word vertically
def remove_vertical(grid, word, row, col):
    for i in range(len(word)):
        grid[row + i][col] = '.'


# Backtracking function
def solve(grid, words, index):
    # All words are placed
    if index == len(words):
        return True

    word = words[index]

    # Try horizontal placement
    for row in range(SIZE):
        for col in range(SIZE):
            if can_place_horizontal(grid, word, row, col):
                place_horizontal(grid, word, row, col)

                if solve(grid, words, index + 1):
                    return True

                # Backtrack
                remove_horizontal(grid, word, row, col)

    # Try vertical placement
    for row in range(SIZE):
        for col in range(SIZE):
            if can_place_vertical(grid, word, row, col):
                place_vertical(grid, word, row, col)

                if solve(grid, words, index + 1):
                    return True

                # Backtrack
                remove_vertical(grid, word, row, col)

    return False


# Display the crossword
def display_grid(grid):
    print("\nGenerated Crossword:\n")

    for row in grid:
        print(" ".join(row))


# Main program
def main():

    words = [
        "PYTHON",
        "JAVA",
        "CODE",
        "PROGRAM",
        "LOOP"
    ]

    # Sort longest words first
    words.sort(key=len, reverse=True)

    grid = create_grid()

    if solve(grid, words, 0):
        display_grid(grid)
    else:
        print("No crossword solution found.")


# Run program
if __name__ == "__main__":
    main()
```
