```python

from abc import ABC, abstractmethod
from typing import List, Optional


# ============================================================
# ABSTRACT BASE CLASS
# ============================================================

class BankAccount(ABC):
    """
    Abstract base class for all bank accounts.
    """

    def __init__(
        self,
        account_number: int,
        account_holder: str,
        balance: float
    ) -> None:
        self.account_number: int = account_number
        self.account_holder: str = account_holder
        self.balance: float = balance

    # --------------------------------------------------------
    # Common deposit method
    # --------------------------------------------------------
    def deposit(self, amount: float) -> None:
        if amount <= 0:
            print("Deposit amount must be greater than 0.")
            return

        self.balance += amount

        print(f"₹{amount:.2f} deposited successfully.")
        print(f"Current balance: ₹{self.balance:.2f}")

    # --------------------------------------------------------
    # Abstract withdrawal method
    # --------------------------------------------------------
    @abstractmethod
    def withdraw(self, amount: float) -> None:
        """
        Child classes must implement this method.
        """
        pass

    # --------------------------------------------------------
    # Display account details
    # --------------------------------------------------------
    def display_details(self) -> None:
        print("\n-----------------------------")
        print("Account Number :", self.account_number)
        print("Account Holder :", self.account_holder)
        print(f"Balance        : ₹{self.balance:.2f}")
        print("-----------------------------")


# ============================================================
# SAVINGS ACCOUNT
# ============================================================

class SavingsAccount(BankAccount):
    """
    Savings account maintains a minimum balance.
    """

    MINIMUM_BALANCE: float = 1000.0

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")

        elif self.balance - amount < self.MINIMUM_BALANCE:
            print(
                f"Withdrawal failed. "
                f"Minimum balance of ₹{self.MINIMUM_BALANCE:.2f} "
                f"must be maintained."
            )

        else:
            self.balance -= amount
            print(f"₹{amount:.2f} withdrawn successfully.")
            print(f"Current balance: ₹{self.balance:.2f}")


# ============================================================
# CURRENT ACCOUNT
# ============================================================

class CurrentAccount(BankAccount):
    """
    Current account allows withdrawal if sufficient
    balance is available.
    """

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")

        elif amount > self.balance:
            print("Withdrawal failed. Insufficient balance.")

        else:
            self.balance -= amount
            print(f"₹{amount:.2f} withdrawn successfully.")
            print(f"Current balance: ₹{self.balance:.2f}")


# ============================================================
# SALARY ACCOUNT
# ============================================================

class SalaryAccount(BankAccount):
    """
    Salary account.
    """

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")

        elif amount > self.balance:
            print("Withdrawal failed. Insufficient balance.")

        else:
            self.balance -= amount
            print(f"₹{amount:.2f} withdrawn successfully.")
            print(f"Current balance: ₹{self.balance:.2f}")


# ============================================================
# BANK MANAGEMENT SYSTEM
# ============================================================

class Bank:
    """
    Manages all bank accounts.
    """

    def __init__(self) -> None:
        self.accounts: List[BankAccount] = []

    # --------------------------------------------------------
    # Add account
    # --------------------------------------------------------
    def add_account(self, account: BankAccount) -> None:
        self.accounts.append(account)

        print(
            f"Account {account.account_number} "
            f"created successfully."
        )

    # --------------------------------------------------------
    # Search account
    # --------------------------------------------------------
    def find_account(
        self,
        account_number: int
    ) -> Optional[BankAccount]:

        for account in self.accounts:
            if account.account_number == account_number:
                return account

        return None

    # --------------------------------------------------------
    # Deposit
    # --------------------------------------------------------
    def deposit_money(
        self,
        account_number: int,
        amount: float
    ) -> None:

        account: Optional[BankAccount] = self.find_account(
            account_number
        )

        if account is None:
            print("Account not found.")
        else:
            account.deposit(amount)

    # --------------------------------------------------------
    # Withdraw
    # --------------------------------------------------------
    def withdraw_money(
        self,
        account_number: int,
        amount: float
    ) -> None:

        account: Optional[BankAccount] = self.find_account(
            account_number
        )

        if account is None:
            print("Account not found.")
        else:
            # Polymorphism:
            # The correct withdraw() method is called
            # according to the account type.
            account.withdraw(amount)

    # --------------------------------------------------------
    # Display one account
    # --------------------------------------------------------
    def display_account(
        self,
        account_number: int
    ) -> None:

        account: Optional[BankAccount] = self.find_account(
            account_number
        )

        if account is None:
            print("Account not found.")
        else:
            account.display_details()

    # --------------------------------------------------------
    # Display all accounts
    # --------------------------------------------------------
    def display_all_accounts(self) -> None:

        if not self.accounts:
            print("No accounts available.")
            return

        print("\n========== ALL ACCOUNTS ==========")

        for account in self.accounts:
            account.display_details()


# ============================================================
# MAIN FUNCTION
# ============================================================

def main() -> None:

    bank: Bank = Bank()

    # --------------------------------------------------------
    # Creating different account objects
    # --------------------------------------------------------

    savings: SavingsAccount = SavingsAccount(
        1001,
        "Rahul",
        10000.0
    )

    current: CurrentAccount = CurrentAccount(
        1002,
        "Priya",
        20000.0
    )

    salary: SalaryAccount = SalaryAccount(
        1003,
        "Arjun",
        30000.0
    )

    # --------------------------------------------------------
    # Add accounts to bank
    # --------------------------------------------------------

    bank.add_account(savings)
    bank.add_account(current)
    bank.add_account(salary)

    # --------------------------------------------------------
    # Deposit operation
    # --------------------------------------------------------

    print("\n========== DEPOSIT ==========")

    bank.deposit_money(1001, 5000.0)

    # --------------------------------------------------------
    # Savings account withdrawal
    # --------------------------------------------------------

    print("\n========== SAVINGS ACCOUNT ==========")

    bank.withdraw_money(1001, 3000.0)

    # --------------------------------------------------------
    # Current account withdrawal
    # --------------------------------------------------------

    print("\n========== CURRENT ACCOUNT ==========")

    bank.withdraw_money(1002, 5000.0)

    # --------------------------------------------------------
    # Salary account withdrawal
    # --------------------------------------------------------

    print("\n========== SALARY ACCOUNT ==========")

    bank.withdraw_money(1003, 10000.0)

    # --------------------------------------------------------
    # Display a single account
    # --------------------------------------------------------

    print("\n========== ACCOUNT DETAILS ==========")

    bank.display_account(1001)

    # --------------------------------------------------------
    # Display all accounts
    # --------------------------------------------------------

    bank.display_all_accounts()


# ============================================================
# PROGRAM START
# ============================================================

if __name__ == "__main__":
    main()



```
