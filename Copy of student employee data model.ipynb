```python

from dataclasses import dataclass
from typing import List


# ============================================================
# PART 1: STUDENT USING TRADITIONAL CLASS
# ============================================================

class TraditionalStudent:
    """
    Student implemented using a traditional Python class.
    """

    def __init__(
        self,
        student_id: int,
        name: str,
        age: int,
        marks: float
    ) -> None:
        self.student_id: int = student_id
        self.name: str = name
        self.age: int = age
        self.marks: float = marks

    def display(self) -> None:
        print(f"Student ID : {self.student_id}")
        print(f"Name       : {self.name}")
        print(f"Age        : {self.age}")
        print(f"Marks      : {self.marks}")

    def calculate_grade(self) -> str:
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        else:
            return "F"


# ============================================================
# PART 2: STUDENT USING DATACLASS
# ============================================================

@dataclass
class Student:
    """
    Student implemented using dataclass.
    """

    student_id: int
    name: str
    age: int
    marks: float

    def display(self) -> None:
        print(f"Student ID : {self.student_id}")
        print(f"Name       : {self.name}")
        print(f"Age        : {self.age}")
        print(f"Marks      : {self.marks}")

    def calculate_grade(self) -> str:
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        else:
            return "F"


# ============================================================
# PART 3: EMPLOYEE USING TRADITIONAL CLASS
# ============================================================

class TraditionalEmployee:
    """
    Employee implemented using a traditional Python class.
    """

    def __init__(
        self,
        employee_id: int,
        name: str,
        department: str,
        salary: float
    ) -> None:
        self.employee_id: int = employee_id
        self.name: str = name
        self.department: str = department
        self.salary: float = salary

    def display(self) -> None:
        print(f"Employee ID : {self.employee_id}")
        print(f"Name        : {self.name}")
        print(f"Department  : {self.department}")
        print(f"Salary      : ₹{self.salary:.2f}")

    def annual_salary(self) -> float:
        return self.salary * 12


# ============================================================
# PART 4: EMPLOYEE USING DATACLASS
# ============================================================

@dataclass
class Employee:
    """
    Employee implemented using dataclass.
    """

    employee_id: int
    name: str
    department: str
    salary: float

    def display(self) -> None:
        print(f"Employee ID : {self.employee_id}")
        print(f"Name        : {self.name}")
        print(f"Department  : {self.department}")
        print(f"Salary      : ₹{self.salary:.2f}")

    def annual_salary(self) -> float:
        return self.salary * 12


# ============================================================
# PART 5: COMPARE STUDENT OBJECTS
# ============================================================

def compare_students() -> None:
    print("\n========================================")
    print("       STUDENT DATA MODEL")
    print("========================================")

    # Traditional class object
    traditional_student: TraditionalStudent = TraditionalStudent(
        101,
        "Rahul",
        20,
        85.5
    )

    # Dataclass object
    dataclass_student: Student = Student(
        101,
        "Rahul",
        20,
        85.5
    )

    print("\n--- Traditional Class Student ---")
    traditional_student.display()
    print("Grade      :", traditional_student.calculate_grade())

    print("\n--- Dataclass Student ---")
    dataclass_student.display()
    print("Grade      :", dataclass_student.calculate_grade())

    print("\n--- Object Representation ---")
    print("Traditional class:")
    print(traditional_student.__dict__)

    print("\nDataclass:")
    print(dataclass_student)


# ============================================================
# PART 6: COMPARE EMPLOYEE OBJECTS
# ============================================================

def compare_employees() -> None:
    print("\n========================================")
    print("       EMPLOYEE DATA MODEL")
    print("========================================")

    # Traditional class object
    traditional_employee: TraditionalEmployee = (
        TraditionalEmployee(
            501,
            "Priya",
            "IT",
            50000.0
        )
    )

    # Dataclass object
    dataclass_employee: Employee = Employee(
        501,
        "Priya",
        "IT",
        50000.0
    )

    print("\n--- Traditional Class Employee ---")
    traditional_employee.display()
    print(
        f"Annual Salary: "
        f"₹{traditional_employee.annual_salary():.2f}"
    )

    print("\n--- Dataclass Employee ---")
    dataclass_employee.display()
    print(
        f"Annual Salary: "
        f"₹{dataclass_employee.annual_salary():.2f}"
    )

    print("\n--- Object Representation ---")
    print("Traditional class:")
    print(traditional_employee.__dict__)

    print("\nDataclass:")
    print(dataclass_employee)


# ============================================================
# PART 7: LIST OF STUDENTS USING DATACLASS
# ============================================================

def display_student_list() -> None:
    print("\n========================================")
    print("       LIST OF STUDENTS")
    print("========================================")

    students: List[Student] = [
        Student(101, "Rahul", 20, 85.5),
        Student(102, "Priya", 21, 92.0),
        Student(103, "Arjun", 20, 76.5),
        Student(104, "Sneha", 22, 64.0)
    ]

    for student in students:
        print(
            f"ID: {student.student_id}, "
            f"Name: {student.name}, "
            f"Marks: {student.marks}, "
            f"Grade: {student.calculate_grade()}"
        )


# ============================================================
# PART 8: LIST OF EMPLOYEES USING DATACLASS
# ============================================================

def display_employee_list() -> None:
    print("\n========================================")
    print("       LIST OF EMPLOYEES")
    print("========================================")

    employees: List[Employee] = [
        Employee(501, "Priya", "IT", 50000.0),
        Employee(502, "Arun", "HR", 45000.0),
        Employee(503, "Meena", "Finance", 60000.0),
        Employee(504, "Kiran", "Marketing", 40000.0)
    ]

    for employee in employees:
        print(
            f"ID: {employee.employee_id}, "
            f"Name: {employee.name}, "
            f"Department: {employee.department}, "
            f"Salary: ₹{employee.salary:.2f}"
        )


# ============================================================
# PART 9: MAIN PROGRAM
# ============================================================

def main() -> None:
    """
    Main function of the program.
    """

    compare_students()

    compare_employees()

    display_student_list()

    display_employee_list()

    print("\n========================================")
    print("             PROGRAM END")
    print("========================================")


# ============================================================
# PROGRAM ENTRY POINT
# ============================================================

if __name__ == "__main__":
    main()
```
