```python


import time
import sys


# ---------------------------------------------------------
# LIST-BASED PROCESSING
# ---------------------------------------------------------

def list_processing(n):
    """
    Creates a list containing squares of numbers from 0 to n-1.
    """
    numbers = [i for i in range(n)]
    squares = [x * x for x in numbers]

    return numbers, squares


# ---------------------------------------------------------
# GENERATOR-BASED PROCESSING
# ---------------------------------------------------------

def generator_processing(n):
    """
    Generates squares one at a time using yield.
    """
    for i in range(n):
        yield i * i


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    # Large dataset
    n = 10_000_000

    print("=" * 60)
    print("LIST VS GENERATOR PROCESSING")
    print("=" * 60)

    # -----------------------------------------------------
    # LIST PROCESSING
    # -----------------------------------------------------

    print("\n1. LIST-BASED PROCESSING")
    print("-" * 60)

    start_time = time.perf_counter()

    numbers, squares = list_processing(n)

    end_time = time.perf_counter()

    list_time = end_time - start_time

    # Memory occupied by list objects
    numbers_memory = sys.getsizeof(numbers)
    squares_memory = sys.getsizeof(squares)

    print("Number of elements:", n)
    print("First 5 numbers:", numbers[:5])
    print("First 5 squares:", squares[:5])
    print("Last 5 squares:", squares[-5:])

    print("Execution time:", round(list_time, 4), "seconds")
    print("Numbers list memory:",
          round(numbers_memory / (1024 * 1024), 2), "MB")
    print("Squares list memory:",
          round(squares_memory / (1024 * 1024), 2), "MB")

    # -----------------------------------------------------
    # GENERATOR PROCESSING
    # -----------------------------------------------------

    print("\n2. GENERATOR-BASED PROCESSING")
    print("-" * 60)

    start_time = time.perf_counter()

    squares_generator = generator_processing(n)

    # Process generator
    total = 0

    for value in squares_generator:
        total += value

    end_time = time.perf_counter()

    generator_time = end_time - start_time

    generator_memory = sys.getsizeof(
        generator_processing(n)
    )

    print("Number of elements:", n)
    print("Generator object memory:",
          round(generator_memory / 1024, 2), "KB")

    print("Total of all squares:", total)

    print("Execution time:",
          round(generator_time, 4), "seconds")

    # -----------------------------------------------------
    # COMPARISON
    # -----------------------------------------------------

    print("\n3. COMPARISON")
    print("=" * 60)

    print("List execution time:",
          round(list_time, 4), "seconds")

    print("Generator execution time:",
          round(generator_time, 4), "seconds")

    print("\nList memory:")
    print("Numbers:", round(numbers_memory / (1024 * 1024), 2), "MB")
    print("Squares:", round(squares_memory / (1024 * 1024), 2), "MB")

    print("\nGenerator memory:",
          round(generator_memory / 1024, 2), "KB")

    print("\nConclusion:")
    print("List stores all values in memory.")
    print("Generator produces values one at a time.")
    print("Therefore, generators are much more memory efficient")
    print("for processing very large datasets.")


# ---------------------------------------------------------
# PROGRAM START
# ---------------------------------------------------------

if __name__ == "__main__":
    main()

```
