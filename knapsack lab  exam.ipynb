```python
def knapsack(weights, values, W):
    n = len(weights)

    # DP table
    dp = [[0] * (W + 1) for _ in range(n + 1)]

    # Fill DP table
    for i in range(1, n + 1):
        for w in range(W + 1):

            if weights[i - 1] <= w:
                dp[i][w] = max(
                    dp[i - 1][w],
                    values[i - 1] + dp[i - 1][w - weights[i - 1]]
                )
            else:
                dp[i][w] = dp[i - 1][w]

    # Find selected projects
    selected = []
    w = W

    for i in range(n, 0, -1):
        if dp[i][w] != dp[i - 1][w]:
            selected.append(i)
            w -= weights[i - 1]

    selected.reverse()

    return dp[n][W], selected


# Project resources
weights = [2, 1, 3, 2]

# Project business values
values = [12, 10, 20, 15]

# Available capacity
W = 5

max_value, projects = knapsack(weights, values, W)

print("Maximum Business Value:", max_value)
print("Selected Projects:", projects)
```
